# SDP-Meta Lab 2: Pipeline Configurations

In the first lab, `LAB 1 - Deploying and Onboarding new Pipelines`, you should have set up the SDP-Meta environment and deployed your first pipeline, which was an APPEND pipline with all records from the Bronze table being appended to the Silver table.

In this lab, we will walk through adding an additional pipeline which is a MERGE pipeline. In a merge pipeline, change events from the Bronze table are merged into the Silver table.



## 1. Configuring the Onboarding files

In a large project with many pipelines, using a single onboarding configuration file to define all of the pipelines can become unwieldy and difficult to maintain. It is recommended to break up the pipelines by domain, or business unit, or another using an additional feature such as a source environment or maintenance schedule, into seperate onboarding configuration files.

In this lab we will show how to use a seperate onboarding file for development or feature testing.

Clone the file conf/onboarding.json to create a new configuration file.
